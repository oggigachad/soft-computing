### 1. Aim
To define and calculate a simple fitness function for evaluating individuals in a Genetic Algorithm using Python.

### 2. Theory: Role of Fitness Function

A **fitness function** $f(x)$ maps each chromosome to a scalar score indicating how good the solution is relative to the optimization objective.

- **Maximization:** Higher $f(x)$ means better fitness (e.g., OneMax problem: $f(x) = \sum_{i=1}^n x_i$).
- **Minimization:** Lower objective value $g(x)$ is better, often converted to fitness via $f(x) = \frac{1}{1 + g(x)}$ (e.g., Sphere function $g(x) = \sum_{i=1}^n x_i^2$).

### 3. Step 1: Implementing Fitness Functions for Binary & Real-Valued Individuals

#### What this cell does:
Evaluates a binary population using the OneMax fitness function and a real-valued population using the Sphere minimization-to-fitness function.

```python
import numpy as np

np.random.seed(42)

def onemax_fitness(individual):
    return int(np.sum(individual))

def sphere_fitness(individual):
    g_x = np.sum(np.square(individual))
    return 1.0 / (1.0 + g_x)

bin_pop = np.random.randint(0, 2, size=(6, 8))
print("Binary Population & OneMax Fitness:")
for idx, ind in enumerate(bin_pop):
    print(f"  Ind {idx+1}: {ind} -> Fitness = {onemax_fitness(ind)}")

real_pop = np.round(np.random.uniform(-3, 3, size=(6, 3)), 3)
print("\nReal-Valued Population & Sphere Fitness:")
for idx, ind in enumerate(real_pop):
    print(f"  Ind {idx+1}: {ind} -> Fitness = {sphere_fitness(ind):.4f}")
```

### 4. Step 2: Plotting Individual Fitness Scores

```python
import matplotlib.pyplot as plt

scores = [onemax_fitness(ind) for ind in bin_pop]
plt.figure(figsize=(7, 4))
plt.bar([f"Ind {i+1}" for i in range(len(bin_pop))], scores, color='teal', edgecolor='black')
plt.title("OneMax Fitness Scores Across Population")
plt.ylabel("Fitness (Number of 1s)")
plt.ylim(0, 9)
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
```
