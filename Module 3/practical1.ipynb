### 2. Theory: Chromosome Encoding in Genetic Algorithms

In a Genetic Algorithm (GA), a candidate solution is represented as a **chromosome** (individual), and a collection of chromosomes forms a **population**. Choosing the right encoding scheme is the first step in designing a GA:

1. **Binary Encoding:** Chromosomes are strings of `0`s and `1`s. Useful for combinatorial problems, knapsack, and feature selection.
2. **Real-Valued (Floating-Point) Encoding:** Genes are continuous real numbers within $[L, U]$. Ideal for continuous function optimization.
3. **Integer Encoding:** Genes are discrete integers. Useful for scheduling, routing, or discrete parameter tuning.
4. **Permutation Encoding:** Chromosomes are ordered permutations of $\{0, 1, \dots, n-1\}$ without repetition. Essential for Traveling Salesman Problem (TSP) and ordering tasks.

### 3. Step 1: Implementing Population Generators for Each Encoding

#### What this cell does:
Defines four functions to generate populations using Binary, Real-Valued, Integer, and Permutation encodings, and prints sample individuals from each.

```python
import numpy as np

np.random.seed(42)

def binary_population(pop_size, chrom_len):
    return np.random.randint(0, 2, size=(pop_size, chrom_len))

def real_population(pop_size, chrom_len, low=-5.0, high=5.0):
    return np.round(np.random.uniform(low, high, size=(pop_size, chrom_len)), 3)

def integer_population(pop_size, chrom_len, low=0, high=10):
    return np.random.randint(low, high + 1, size=(pop_size, chrom_len))

def permutation_population(pop_size, chrom_len):
    return np.array([np.random.permutation(chrom_len) for _ in range(pop_size)])

pop_size, chrom_len = 5, 6

print("1. Binary Encoding Population:\n", binary_population(pop_size, chrom_len))
print("\n2. Real-Valued Encoding Population:\n", real_population(pop_size, chrom_len))
print("\n3. Integer Encoding Population:\n", integer_population(pop_size, chrom_len))
print("\n4. Permutation Encoding Population:\n", permutation_population(pop_size, chrom_len))
```

### 4. Step 2: Visualizing Population Distributions

#### What this cell does:
Visualizes gene distributions for Binary, Real-Valued, and Integer populations using Matplotlib.

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(1, 3, figsize=(14, 4))

bin_pop = binary_population(50, 10)
axes[0].bar(['0', '1'], [np.sum(bin_pop == 0), np.sum(bin_pop == 1)], color=['steelblue', 'coral'])
axes[0].set_title("Binary Gene Counts")
axes[0].set_ylabel("Frequency")

real_pop = real_population(100, 5)
axes[1].hist(real_pop.flatten(), bins=15, color='seagreen', edgecolor='black')
axes[1].set_title("Real-Valued Gene Distribution")
axes[1].set_xlabel("Value")

int_pop = integer_population(100, 5, 0, 9)
vals, counts = np.unique(int_pop, return_counts=True)
axes[2].bar(vals, counts, color='mediumpurple', edgecolor='black')
axes[2].set_title("Integer Gene Distribution")
axes[2].set_xlabel("Integer Value")

plt.tight_layout()
plt.show()
```
