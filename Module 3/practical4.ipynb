### 2. Theory: Crossover (Recombination)

Crossover combines genetic material from two parent chromosomes to produce offspring:
1. **Single-Point Crossover:** Splits both parents at one random point and swaps tails.
2. **Two-Point Crossover:** Splits parents at two random points and swaps the middle segment.
3. **Uniform Crossover:** Each gene is independently swapped between parents with probability $0.5$.

### 3. Step 1: Implementing Single-Point, Two-Point, and Uniform Crossover

```python
import numpy as np

np.random.seed(42)

def single_point_crossover(p1, p2):
    pt = np.random.randint(1, len(p1))
    c1 = np.concatenate([p1[:pt], p2[pt:]])
    c2 = np.concatenate([p2[:pt], p1[pt:]])
    return c1, c2, pt

def two_point_crossover(p1, p2):
    pts = sorted(np.random.choice(range(1, len(p1)), size=2, replace=False))
    c1 = np.concatenate([p1[:pts[0]], p2[pts[0]:pts[1]], p1[pts[1]:]])
    c2 = np.concatenate([p2[:pts[0]], p1[pts[0]:pts[1]], p2[pts[1]:]])
    return c1, c2, pts

def uniform_crossover(p1, p2, prob=0.5):
    mask = np.random.rand(len(p1)) < prob
    c1 = np.where(mask, p2, p1)
    c2 = np.where(mask, p1, p2)
    return c1, c2, mask

p1 = np.array([1, 1, 1, 1, 0, 0, 0, 0])
p2 = np.array([0, 0, 0, 0, 1, 1, 1, 1])

print("Parent 1:", p1)
print("Parent 2:", p2)

c1, c2, pt = single_point_crossover(p1, p2)
print(f"\nSingle-Point (pt={pt}):\n  Child 1: {c1}\n  Child 2: {c2}")

c1, c2, pts = two_point_crossover(p1, p2)
print(f"\nTwo-Point (pts={pts}):\n  Child 1: {c1}\n  Child 2: {c2}")

c1, c2, _ = uniform_crossover(p1, p2)
print(f"\nUniform Crossover:\n  Child 1: {c1}\n  Child 2: {c2}")
```

### 4. Conclusion
We implemented Single-Point, Two-Point, and Uniform crossover operators to generate offspring from parent chromosomes.
