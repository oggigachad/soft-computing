### 2. Theory: Complete Genetic Algorithm Loop

A Simple Genetic Algorithm (SGA) evolves a population over $G$ generations by repeating:
1. **Initialization:** Create random initial population.
2. **Evaluation:** Compute fitness of each individual.
3. **Selection:** Choose parents (e.g., Tournament Selection).
4. **Crossover:** Recombine pairs of parents with probability $p_c$.
5. **Mutation:** Mutate offspring with probability $p_m$.
6. **Replacement:** Form the next generation (optionally keeping the best individual via **Elitism**).

### 3. Step 1: Implementing Simple GA on the OneMax Problem

```python
import numpy as np

np.random.seed(42)

def simple_ga(pop_size=20, chrom_len=16, generations=30, pc=0.8, pm=0.05):
    pop = np.random.randint(0, 2, size=(pop_size, chrom_len))
    best_hist = []

    for gen in range(generations):
        fitness = np.sum(pop, axis=1)
        best_idx = np.argmax(fitness)
        best_hist.append(fitness[best_idx])

        if fitness[best_idx] == chrom_len:
            print(f"Optimal solution found at generation {gen}!")
            break

        new_pop = [pop[best_idx].copy()]  # Elitism: keep best
        while len(new_pop) < pop_size:
            # Tournament selection (k=3)
            c1 = np.random.choice(pop_size, 3, replace=False)
            p1 = pop[c1[np.argmax(fitness[c1])]]
            c2 = np.random.choice(pop_size, 3, replace=False)
            p2 = pop[c2[np.argmax(fitness[c2])]]

            # Single-point crossover
            if np.random.rand() < pc:
                pt = np.random.randint(1, chrom_len)
                o1 = np.concatenate([p1[:pt], p2[pt:]])
                o2 = np.concatenate([p2[:pt], p1[pt:]])
            else:
                o1, o2 = p1.copy(), p2.copy()

            # Bit-flip mutation
            for o in (o1, o2):
                mask = np.random.rand(chrom_len) < pm
                o[mask] = 1 - o[mask]
                if len(new_pop) < pop_size:
                    new_pop.append(o)

        pop = np.array(new_pop)

    final_fit = np.sum(pop, axis=1)
    best_ind = pop[np.argmax(final_fit)]
    return best_ind, np.max(final_fit), best_hist

best_ind, best_fit, history = simple_ga()
print("Best Individual:", best_ind)
print("Best Fitness:   ", best_fit)
```

### 4. Conclusion
We combined selection, crossover, mutation, and elitism into a working Simple Genetic Algorithm that solves the OneMax problem.
