### 2. Theory: Mutation

Mutation introduces small random changes into chromosomes to maintain genetic diversity and prevent premature convergence:
1. **Bit-Flip Mutation (Binary):** Flips each bit (`0` $\leftrightarrow$ `1`) with probability $p_m$.
2. **Gaussian Mutation (Real-Valued):** Adds $\mathcal{N}(0, \sigma^2)$ noise to genes with probability $p_m$.
3. **Swap Mutation (Permutation):** Randomly swaps two positions in a permutation chromosome.

### 3. Step 1: Implementing Bit-Flip, Gaussian, and Swap Mutation

```python
import numpy as np

np.random.seed(42)

def bit_flip_mutation(ind, pm=0.2):
    mutated = ind.copy()
    for i in range(len(mutated)):
        if np.random.rand() < pm:
            mutated[i] = 1 - mutated[i]
    return mutated

def gaussian_mutation(ind, pm=0.3, sigma=0.5):
    mutated = ind.copy().astype(float)
    for i in range(len(mutated)):
        if np.random.rand() < pm:
            mutated[i] += np.random.normal(0, sigma)
    return np.round(mutated, 3)

def swap_mutation(ind):
    mutated = ind.copy()
    i, j = np.random.choice(len(mutated), size=2, replace=False)
    mutated[i], mutated[j] = mutated[j], mutated[i]
    return mutated, (i, j)

b_ind = np.array([1, 0, 1, 1, 0, 0, 1, 0])
print("Binary Original:", b_ind)
print("Bit-Flip Mutated:", bit_flip_mutation(b_ind, pm=0.25))

r_ind = np.array([1.2, -0.5, 3.1, 0.0, -2.4])
print("\nReal Original:", r_ind)
print("Gaussian Mutated:", gaussian_mutation(r_ind, pm=0.4, sigma=0.5))

p_ind = np.array([0, 1, 2, 3, 4, 5])
p_mut, swapped = swap_mutation(p_ind)
print("\nPermutation Original:", p_ind)
print(f"Swap Mutated (swapped {swapped}):", p_mut)
```

### 4. Conclusion
We implemented Bit-Flip, Gaussian, and Swap mutation operators for binary, real-valued, and permutation chromosomes.
